```python
# -*- coding: utf-8 -*-
"""
FRED API
@author: adam
"""

# Import modules
import requests
import pandas as pd

# ✅ Define your API key (must be a string)
api_key = "cf97255b0c0b736605b5d9e267e2977e"

# Define the FRED API endpoint
base_url = "https://api.stlouisfed.org/fred/"

# ============================================================
# Get observation data
# ============================================================
obs_endpoint = "series/observations"
series_id = "CPIAUCSL"
start_date = "2000-01-01"
end_date = "2023-06-30"

obs_params = {
    "series_id": series_id,
    "api_key": api_key,
    "file_type": "json",
    "observation_start": start_date,
    "observation_end": end_date,
}

response = requests.get(base_url + obs_endpoint, params=obs_params)

if response.status_code == 200:
    res_data = response.json()
    obs_data = pd.DataFrame(res_data["observations"])
    obs_data["date"] = pd.to_datetime(obs_data["date"])
    obs_data.set_index("date", inplace=True)
    obs_data["value"] = obs_data["value"].astype(float)
    print(obs_data.head())
else:
    print("Failed to retrieve observations:", response.status_code)

# ============================================================
# Get category details
# ============================================================
cat_endpoint = "category"
cat_params = {"api_key": api_key, "file_type": "json", "category_id": 0}

response = requests.get(base_url + cat_endpoint, params=cat_params)
if response.status_code == 200:
    res_data = response.json()
    print("Category name:", res_data["categories"][0]["name"])
else:
    print("Failed to retrieve category details:", response.status_code)

# ============================================================
# Get category children
# ============================================================
child_endpoint = "category/children"
child_params = {"api_key": api_key, "file_type": "json", "category_id": 0}

response = requests.get(base_url + child_endpoint, params=child_params)
if response.status_code == 200:
    res_data = response.json()
    print("Child categories:", [c["name"] for c in res_data["categories"]])
else:
    print("Failed to retrieve category children:", response.status_code)

# ============================================================
# Get series from category
# ============================================================
cat_srs_endpoint = "category/series"
cat_srs_params = {"api_key": api_key, "file_type": "json", "category_id": 32455}

response = requests.get(base_url + cat_srs_endpoint, params=cat_srs_params)
if response.status_code == 200:
    cat_srs_data = response.json()
    cat_srs = pd.DataFrame(cat_srs_data["seriess"])
    cat_srs = cat_srs.sort_values(by="popularity", ascending=False, ignore_index=True)
    print(cat_srs[["id", "title"]].head())
else:
    print("Failed to retrieve category series:", response.status_code)

# ============================================================
# Get releases
# ============================================================
rel_id_endpoint = "releases"
rel_id_params = {"api_key": api_key, "file_type": "json"}

response = requests.get(base_url + rel_id_endpoint, params=rel_id_params)
if response.status_code == 200:
    rel_data = response.json()
    releases = pd.DataFrame(rel_data["releases"])
    print(releases[["id", "name"]].head())
else:
    print("Failed to retrieve releases:", response.status_code)

# ============================================================
# Get release series
# ============================================================
rel_srs_endpoint = "release/series"
rel_srs_params = {"api_key": api_key, "file_type": "json", "release_id": 10}

response = requests.get(base_url + rel_srs_endpoint, params=rel_srs_params)
if response.status_code == 200:
    rel_srs_data = response.json()
    rel_srs = pd.DataFrame(rel_srs_data["seriess"])
    print(rel_srs[["id", "title"]].head())
else:
    print("Failed to retrieve release series:", response.status_code)

# ============================================================
# Get sources
# ============================================================
src_endpoint = "sources"
src_params = {"api_key": api_key, "file_type": "json"}

response = requests.get(base_url + src_endpoint, params=src_params)
if response.status_code == 200:
    src_data = response.json()
    src = pd.DataFrame(src_data["sources"])
    print(src[["id", "name"]].head())
else:
    print("Failed to retrieve sources:", response.status_code)
```

####| Variable name | What it contains                           |
####| ------------- | ------------------------------------------ |
####| `obs_data`    | CPI time series data (dates & values)      |
####| `cat_srs`     | List of series from a given category       |
####| `releases`    | List of official data releases             |
####| `rel_srs`     | Series that belong to a particular release |
####| `src`         | Data sources list                          |
